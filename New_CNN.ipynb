```python
# ============================================================
# BUILDING DAMAGE CLASSIFICATION USING SATELLITE IMAGES
# DATA LOADING + CNN MODEL + TRAINING + EVALUATION + SAVE MODEL
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from pathlib import Path

from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing.image import ImageDataGenerator

from tensorflow.keras.layers import (
    Conv2D,
    MaxPool2D,
    Dense,
    Flatten
)

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)


# ============================================================
# 2. DATASET PATHS
# ============================================================

# Your original Kaggle dataset paths

train_directory = Path(
    "/content/train_another"
)

validation_directory = Path(
    "/content/validation_another"
)

unbalanced_test_directory = Path(
    "/content/test_another"
)

balanced_test_directory = Path(
    "/content/test"
)


# Print paths
print("Training Path:")
print(train_directory)

print("\nValidation Path:")
print(validation_directory)

print("\nUnbalanced Test Path:")
print(unbalanced_test_directory)

print("\nBalanced Test Path:")
print(balanced_test_directory)


# ============================================================
# 3. CHECK DATASET PATHS
# ============================================================

print("\nChecking dataset folders...")

print(
    "Train Exists:",
    train_directory.exists()
)

print(
    "Validation Exists:",
    validation_directory.exists()
)

print(
    "Unbalanced Test Exists:",
    unbalanced_test_directory.exists()
)

print(
    "Balanced Test Exists:",
    balanced_test_directory.exists()
)


# ============================================================
# 4. CHECK CLASS FOLDERS
# ============================================================

print("\nTraining folders:")

if train_directory.exists():

    print(
        os.listdir(train_directory)
    )


# Expected structure:
#
# train_another/
# ├── damage/
# └── no_damage/
#
# validation_another/
# ├── damage/
# └── no_damage/
#
# test_another/
# ├── damage/
# └── no_damage/
#
# test/
# ├── damage/
# └── no_damage/


# ============================================================
# 5. IMAGE SETTINGS
# ============================================================

IMAGE_HEIGHT = 128
IMAGE_WIDTH = 128

IMAGE_SIZE = (
    IMAGE_HEIGHT,
    IMAGE_WIDTH
)

BATCH_SIZE = 100


# ============================================================
# 6. DATA AUGMENTATION
# ============================================================

train_gen = ImageDataGenerator(

    rotation_range=10,

    width_shift_range=0.2,

    height_shift_range=0.2,

    zoom_range=0.2,

    horizontal_flip=True,

    rescale=1/255.0,

    brightness_range=[0.2, 1.2]
)


validation_gen = ImageDataGenerator(

    rotation_range=10,

    width_shift_range=0.2,

    height_shift_range=0.2,

    zoom_range=0.2,

    horizontal_flip=True,

    rescale=1/255.0,

    brightness_range=[0.2, 1.2]
)


# Keeping this same as your notebook
test_unbalanced_gen = ImageDataGenerator()

test_balanced_gen = ImageDataGenerator()


# ============================================================
# 7. LOAD TRAINING DATA
# ============================================================

train_data = train_gen.flow_from_directory(

    directory=train_directory,

    target_size=IMAGE_SIZE,

    class_mode="binary",

    color_mode="rgb",

    shuffle=True,

    batch_size=BATCH_SIZE
)


# ============================================================
# 8. LOAD VALIDATION DATA
# ============================================================

val_data = validation_gen.flow_from_directory(

    directory=validation_directory,

    target_size=IMAGE_SIZE,

    class_mode="binary",

    color_mode="rgb",

    shuffle=True,

    batch_size=BATCH_SIZE
)


# ============================================================
# 9. LOAD UNBALANCED TEST DATA
# ============================================================

unbalanced_data = test_unbalanced_gen.flow_from_directory(

    directory=unbalanced_test_directory,

    target_size=IMAGE_SIZE,

    class_mode="binary",

    color_mode="rgb",

    shuffle=False,

    batch_size=BATCH_SIZE
)


# ============================================================
# 10. LOAD BALANCED TEST DATA
# ============================================================

balanced_data = test_balanced_gen.flow_from_directory(

    directory=balanced_test_directory,

    target_size=IMAGE_SIZE,

    class_mode="binary",

    color_mode="rgb",

    shuffle=False,

    batch_size=BATCH_SIZE
)


# ============================================================
# 11. DISPLAY CLASS MAPPING
# ============================================================

print("\nClass Mapping:")

print(
    train_data.class_indices
)


# Normally:
#
# damage     -> 0
# no_damage  -> 1


# ============================================================
# 12. DISPLAY DATASET INFORMATION
# ============================================================

print("\nDataset Information")

print(
    "Training Images:",
    train_data.samples
)

print(
    "Validation Images:",
    val_data.samples
)

print(
    "Unbalanced Test Images:",
    unbalanced_data.samples
)

print(
    "Balanced Test Images:",
    balanced_data.samples
)


# ============================================================
# 13. BUILD CNN MODEL
# ============================================================

model = Sequential(
    name="Base_Model"
)


# -----------------------------
# Convolution Block 1
# -----------------------------

model.add(
    Conv2D(
        32,
        kernel_size=(3, 3),
        activation="relu",
        input_shape=(128, 128, 3)
    )
)


model.add(
    Conv2D(
        32,
        kernel_size=(3, 3),
        activation="relu"
    )
)


model.add(
    MaxPool2D(
        pool_size=(2, 2)
    )
)


# -----------------------------
# Convolution Block 2
# -----------------------------

model.add(
    Conv2D(
        64,
        kernel_size=(3, 3),
        activation="relu"
    )
)


model.add(
    Conv2D(
        64,
        kernel_size=(3, 3),
        activation="relu"
    )
)


model.add(
    MaxPool2D(
        pool_size=(2, 2)
    )
)


# -----------------------------
# Convolution Block 3
# -----------------------------

model.add(
    Conv2D(
        128,
        kernel_size=(3, 3),
        activation="relu"
    )
)


model.add(
    Conv2D(
        128,
        kernel_size=(3, 3),
        activation="relu"
    )
)


model.add(
    MaxPool2D(
        pool_size=(2, 2)
    )
)


# ============================================================
# 14. FLATTEN
# ============================================================

model.add(
    Flatten()
)


# ============================================================
# 15. FULLY CONNECTED LAYERS
# ============================================================

model.add(
    Dense(
        1024,
        activation="relu"
    )
)


model.add(
    Dense(
        1024,
        activation="relu"
    )
)


# ============================================================
# 16. OUTPUT LAYER
# ============================================================

model.add(
    Dense(
        1,
        activation="sigmoid"
    )
)


# ============================================================
# 17. MODEL SUMMARY
# ============================================================

print("\nCNN Model Architecture:")

model.summary()


# ============================================================
# 18. COMPILE MODEL
# ============================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-4
    ),

    loss="binary_crossentropy",

    metrics=["accuracy"]
)


print("\nModel compiled successfully.")


# ============================================================
# 19. TRAIN MODEL
# ============================================================

EPOCHS = 10


history = model.fit(

    train_data,

    validation_data=val_data,

    epochs=EPOCHS
)


# ============================================================
# 20. SAVE TRAINED MODEL
# ============================================================

MODEL_PATH = "/content/working/building_damage_cnn_model.keras"


model.save(
    MODEL_PATH
)


print("\nModel successfully saved.")

print(
    "Saved Model Path:",
    MODEL_PATH
)


# ============================================================
# 21. EVALUATE BALANCED TEST DATASET
# ============================================================

balanced_loss, balanced_accuracy = model.evaluate(

    balanced_data,

    verbose=1
)


print(
    "\nBalanced Test Loss:",
    balanced_loss
)

print(
    "Balanced Test Accuracy:",
    balanced_accuracy
)

print(
    f"Balanced Accuracy: "
    f"{balanced_accuracy * 100:.2f}%"
)


# ============================================================
# 22. EVALUATE UNBALANCED TEST DATASET
# ============================================================

unbalanced_loss, unbalanced_accuracy = model.evaluate(

    unbalanced_data,

    verbose=1
)


print(
    "\nUnbalanced Test Loss:",
    unbalanced_loss
)

print(
    "Unbalanced Test Accuracy:",
    unbalanced_accuracy
)

print(
    f"Unbalanced Accuracy: "
    f"{unbalanced_accuracy * 100:.2f}%"
)


# ============================================================
# 23. TRAINING ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title(
    "CNN Training and Validation Accuracy"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 24. TRAINING LOSS GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title(
    "CNN Training and Validation Loss"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 25. PREDICTIONS ON BALANCED TEST DATA
# ============================================================

balanced_data.reset()


balanced_probabilities = model.predict(
    balanced_data
)


balanced_predictions = (
    balanced_probabilities >= 0.5
).astype(int).flatten()


balanced_true_labels = (
    balanced_data.classes
)


# ============================================================
# 26. CLASSIFICATION REPORT
# ============================================================

print(
    "\nClassification Report "
    "- Balanced Test Dataset"
)


print(
    classification_report(
        balanced_true_labels,
        balanced_predictions,
        target_names=[
            "damage",
            "no_damage"
        ]
    )
)


# ============================================================
# 27. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    balanced_true_labels,

    balanced_predictions
)


print(
    "\nConfusion Matrix:"
)

print(cm)


# ============================================================
# 28. PRECISION, RECALL, F1 AND ROC-AUC
# ============================================================

precision = precision_score(

    balanced_true_labels,

    balanced_predictions
)


recall = recall_score(

    balanced_true_labels,

    balanced_predictions
)


f1 = f1_score(

    balanced_true_labels,

    balanced_predictions
)


roc_auc = roc_auc_score(

    balanced_true_labels,

    balanced_probabilities.flatten()
)


print(
    "\nPrecision:",
    precision
)

print(
    "Recall:",
    recall
)

print(
    "F1 Score:",
    f1
)

print(
    "ROC-AUC:",
    roc_auc
)


# ============================================================
# 29. FINAL RESULT
# ============================================================

print("\n=================================")

print("MODEL TRAINING COMPLETED")

print("=================================")

print(
    f"Balanced Test Accuracy   : "
    f"{balanced_accuracy * 100:.2f}%"
)

print(
    f"Unbalanced Test Accuracy : "
    f"{unbalanced_accuracy * 100:.2f}%"
)

print(
    f"Precision                : "
    f"{precision:.4f}"
)

print(
    f"Recall                   : "
    f"{recall:.4f}"
)

print(
    f"F1 Score                 : "
    f"{f1:.4f}"
)

print(
    f"ROC-AUC                  : "
    f"{roc_auc:.4f}"
)

print(
    "\nModel Saved At:"
)

print(MODEL_PATH)

print("=================================")
```
